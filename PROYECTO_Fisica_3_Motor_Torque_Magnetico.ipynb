## 2. Introducción
Un motor eléctrico convierte energía eléctrica en energía mecánica mediante la interacción entre una corriente y un campo magnético. En el motor casero de este proyecto, una bobina por la que circula corriente se ubica en el campo de un imán: sobre ella aparece un torque que tiende a alinear su momento magnético con el campo. Como el conmutador interrumpe la corriente durante media vuelta, la bobina sigue girando por inercia y el impulso se repite en cada vuelta.

## 4. Teoría
El principio fundamental detrás de un motor de conducción magnética es la **Fuerza de Lorentz**. Cuando una carga eléctrica se mueve dentro de un campo magnético, experimenta una fuerza perpendicular tanto a su velocidad como al campo. A nivel macroscópico, un alambre conductor de longitud $\vec{L}$ por el que circula una corriente $I$, inmerso en un campo magnético $\vec{B}$, experimenta una fuerza magnética expresada como:

$$\vec{F} = I (\vec{L} \times \vec{B})$$

Esta fuerza, al estar aplicada a cierta distancia del eje de rotación, produce un torque ($\vec{\tau} = \vec{r} \times \vec{F}$) que es el responsable de mantener la estructura en un movimiento rotacional continuo mientras el circuito se encuentre cerrado.

A continuación se presentan los conceptos de la teoría electromagnética que se usaron en el proyecto.

### 4.1 Campo magnético
**Imán.** Sobre la bobina se supone un campo del imán aproximadamente uniforme, dirigido hacia arriba: $\vec B=(0,0,B)$.

**Bobina (ley de Biot-Savart).** La bobina, al transportar corriente, genera su propio campo. Cada elemento de alambre $d\vec l$ aporta:

$$d\vec B=\frac{\mu_0}{4\pi}\,\frac{N\,I\,d\vec l\times\hat r}{r^2}\quad\Rightarrow\quad \vec B(\vec r)=\frac{\mu_0 N I}{4\pi}\oint\frac{d\vec l\times\vec R}{|\vec R|^3}$$

En el centro de la espira y sobre su eje, el resultado es:

$$B_{centro}=\frac{\mu_0 N I}{2r},\qquad B(z)=\frac{\mu_0 N I r^2}{2\,(r^2+z^2)^{3/2}}$$

El campo total es la suma (principio de superposición) del campo del imán y el de la bobina. Esta ley es la misma que se usa en el código de Algoritmos Computacionales para el cable coaxial.

### 4.2 Corriente y campo eléctrico en el conductor
Con $A_w=\pi d^2/4$ la sección del alambre, la densidad de corriente y el campo eléctrico en el conductor son:

$$J=\frac{I}{A_w},\qquad E=\rho\,J,\qquad v_d=\frac{J}{n\,e}$$

donde $\rho$ es la resistividad, $n$ la densidad de electrones libres y $v_d$ la velocidad de arrastre. La resistencia de la bobina depende de su longitud $\ell=N\,2\pi r$ y de la temperatura:

$$R_{bob}(T)=\frac{\rho\,\ell}{A_w}\,[1+\alpha\,(T-20)],\qquad R=R_{bob}+R_{ext}$$

Además, cuando el alambre se mueve con velocidad $\vec v$ dentro del campo $\vec B$, sus cargas sienten un campo eléctrico efectivo $\vec E_{mov}=\vec v\times\vec B$.

### 4.3 Flujo magnético y fuerza electromotriz (ley de Faraday-Lenz)
El flujo magnético a través de la bobina de $N$ espiras y área $A=\pi r^2$ depende de su orientación $\theta$:

$$\Phi=N\,\vec B\cdot\vec A=-N A B\cos\theta$$

Al girar con velocidad angular $\omega=\dot\theta$, el flujo cambia y se induce una fem:

$$\varepsilon=-\frac{d\Phi}{dt}\;\Rightarrow\;|\varepsilon|=N A B\,\omega\,\sin\theta$$

Por la ley de Lenz, esta fem se opone a la causa que la produce, es decir, actúa contra el voltaje $V$ de la fuente. La corriente se obtiene de la ley de Ohm con ambas fem:

$$I=\frac{V-\varepsilon}{R}$$

Mientras más rápido gira el motor, mayor es $\varepsilon$ y menor es la corriente. La constante $K=NAB$ resume cuánto influyen las espiras, el área y el campo.

### 4.4 Momento magnético y torque
Una bobina con corriente se comporta como un dipolo magnético con momento $\vec\mu=N\,I\,A\,\hat n$, donde $\hat n=(0,\sin\theta,-\cos\theta)$ es la normal al plano de la bobina. En el campo del imán aparece un torque:

$$\vec\tau=\vec\mu\times\vec B\;\Rightarrow\;\tau_x=N\,I\,A\,B\,\sin\theta$$

Sobre cada elemento de alambre actúa $d\vec F=N\,I\,d\vec l\times\vec B$; como el lazo es cerrado y el campo es uniforme, la fuerza neta es cero ($\vec F_{neta}=\vec 0$): la bobina **solo gira**, no se traslada.

### 4.5 Conmutador
El alambre pelado solo por la mitad en uno de los extremos cierra el circuito durante media vuelta. Por eso la corriente es:

$$I=\begin{cases}\dfrac{V-\varepsilon}{R} & 0\le\theta \bmod 2\pi<\pi\\[2mm] 0 & \pi\le\theta \bmod 2\pi<2\pi\end{cases}$$

En la mitad sin corriente no hay torque y la bobina continúa por inercia, lo que mantiene el sentido de giro.

### 4.6 Dinámica de rotación
Con $J$ el momento de inercia y $b$ la fricción viscosa, el movimiento se describe con:

$$J\,\dot\omega=\tau_x-b\,\omega,\qquad \dot\theta=\omega$$

Estas ecuaciones se integran numéricamente con el método de Runge-Kutta de cuarto orden. Si el giro es rápido, el torque medio es $\langle\tau\rangle=\dfrac{KV}{\pi R}-\dfrac{K^2\omega}{4R}$ y, igualándolo a $b\,\omega$, se obtiene la velocidad terminal:

$$\omega_{ss}=\frac{KV/(\pi R)}{\,b+K^2/(4R)\,},\qquad \omega_0=\frac{4V}{\pi K}\ \ (b=0)$$

### 4.7 Energía y calentamiento
La potencia entregada por la fuente se reparte entre el calor por efecto Joule y el trabajo mecánico:

$$P_{fuente}=V\langle I\rangle=\langle I^2R\rangle+\langle\tau\,\omega\rangle$$

El alambre se calienta con el efecto Joule y se enfría por convección con el aire:

$$\frac{dT}{dt}=\frac{I^2R_{bob}-h\,S\,(T-T_{amb})}{m\,c}$$

donde $m$ es la masa del cobre, $c$ su calor específico, $S$ la superficie de enfriamiento y $h$ el coeficiente de convección.

### Planteamiento del modelo físico
Aquí están las constantes, los datos del motor y las fórmulas que usamos: cuánta corriente pasa, cuánto torque se genera, cómo gira y se calienta, y cómo se calcula el campo magnético de la bobina.

```python
# MODELO FÍSICO: parámetros, integración RK4, Biot-Savart y forma cerrada
import math
from dataclasses import dataclass, replace
import numpy as np

# Constantes que se usan en todo el código
PI = math.pi
DOS_PI = 2.0 * math.pi

# Constantes físicas y del material
MU0 = 4e-7 * PI        # permeabilidad del vacío [T·m/A]
RHO = 1.68e-8          # resistividad del cobre a 20 °C [Ω·m]
ALFA_T = 0.0039        # coeficiente de temperatura del cobre [1/K]
DENS = 8960.0          # densidad del cobre [kg/m³]
CP = 385.0             # calor específico del cobre [J/(kg·K)]
H_CONV = 10.0          # convección natural [W/(m²·K)]
T_AMB = 20.0           # temperatura ambiente [°C]
G = 9.81               # gravedad [m/s²]
N_E = 8.5e28           # electrones libres del cobre [1/m³]
Q_E = 1.602e-19        # carga elemental [C]
J_AXLE = 2e-7          # inercia del eje y contactos [kg·m²]
Z0 = 0.026             # altura del eje sobre el imán [m]
X_SOP = 0.055          # posición de los soportes en X [m]
R_IMAN = 0.045         # radio del imán [m]
T_ESMALTE = 155.0      # límite térmico del esmalte [°C]


# Los datos que se pueden cambiar del motor. Valores iniciales: 6 espiras, 9 V, radio 1,5 cm y 0,4 Ω.
@dataclass
class Params:
    """Parámetros editables del motor"""
    N: int = 6             # espiras
    V: float = 9.0         # voltaje de la fuente [V]
    B: float = 0.030       # campo del imán [T]
    b: float = 5e-6        # fricción viscosa [N·m·s]
    d: float = 0.5e-3      # diámetro del alambre [m]
    r: float = 0.015       # radio de la bobina [m]
    Rext: float = 0.4      # resistencia externa (pila, clips, contactos) [Ω]


# Calcula área, longitud del alambre, resistencia, masa e inercia a partir de N, radio y grosor.
def geometria(p: Params) -> dict:
    """Magnitudes derivadas de la geometría del alambre."""
    A = PI * p.r ** 2                 # área de una espira
    lt = 2 * PI * p.r                 # longitud de una espira
    l = p.N * lt                      # longitud total
    Aw = PI * p.d ** 2 / 4            # sección del alambre
    R0 = RHO * l / Aw                 # resistencia a 20 °C
    m = DENS * l * Aw                 # masa de cobre
    J = J_AXLE + 0.5 * m * p.r ** 2   # inercia (anillos delgados sobre su diámetro: ½ m r²)
    S = l * PI * p.d                  # superficie de enfriamiento
    return dict(A=A, lt=lt, l=l, Aw=Aw, R0=R0, m=m, J=J, S=S, Cth=m * CP, K0=p.N * A * p.B)


# Dice si el conmutador deja pasar corriente (solo media vuelta, como la mitad pelada del alambre).
def conmutador(th: float) -> bool:
    """Lógica estricta del conmutador: hay corriente solo si 0 <= theta mod 2π < π."""
    return (th % DOS_PI) < PI


# Simula el giro paso a paso (método Runge-Kutta): calcula ángulo, velocidad, corriente, torque y temperatura en cada instante.
def simular(p: Params, t_total: float = 1.5, dt: float = 2e-4,
            theta0: float = 0.4, omega0: float = 0.0) -> dict:
    """Integra θ, ω y la temperatura con Runge-Kutta 4. Devuelve series temporales."""
    g = geometria(p)
    K0, R0, Rext, V, b, J = g['K0'], g['R0'], p.Rext, p.V, p.b, g['J']
    hS, Cth = H_CONV * g['S'], g['Cth']

    def ele(th, om, T):
        """Estado eléctrico: (I, ε, τ, P_joule_bobina)."""
        if (th % DOS_PI) < PI:
            Rb = R0 * (1 + ALFA_T * (T - T_AMB))
            s = math.sin(th)
            eps = K0 * om * s
            I = (V - eps) / (Rb + Rext)
            return I, eps, K0 * s * I, I * I * Rb
        return 0.0, 0.0, 0.0, 0.0

    def deriv(th, om, T):
        I, eps, tau, pj = ele(th, om, T)
        return om, (tau - b * om) / J, (pj - hS * (T - T_AMB)) / Cth

    n = int(round(t_total / dt))
    t = np.arange(n + 1) * dt
    TH = np.empty(n + 1); OM = np.empty(n + 1); TE = np.empty(n + 1)
    II = np.empty(n + 1); TAU = np.empty(n + 1); EPS = np.empty(n + 1)
    th, om, T = theta0, omega0, T_AMB
    for k in range(n + 1):
        TH[k], OM[k], TE[k] = th, om, T
        II[k], EPS[k], TAU[k], _ = ele(th, om, T)
        if k == n:
            break
        a = deriv(th, om, T)
        bb = deriv(th + 0.5 * dt * a[0], om + 0.5 * dt * a[1], T + 0.5 * dt * a[2])
        c = deriv(th + 0.5 * dt * bb[0], om + 0.5 * dt * bb[1], T + 0.5 * dt * bb[2])
        d = deriv(th + dt * c[0], om + dt * c[1], T + dt * c[2])
        th += dt * (a[0] + 2 * bb[0] + 2 * c[0] + d[0]) / 6
        om += dt * (a[1] + 2 * bb[1] + 2 * c[1] + d[1]) / 6
        T = min(1500.0, T + dt * (a[2] + 2 * bb[2] + 2 * c[2] + d[2]) / 6)
    return dict(t=t, th=TH, om=OM, T=TE, I=II, eps=EPS, tau=TAU, g=g, p=p)


# Dibuja la bobina como muchos segmentos pequeños para poder calcular su campo magnético.
def lazo(th: float, r: float, M: int = 48):
    """Puntos medios y elementos dl del lazo. El recorrido (φ decreciente) da μ ∥ n̂."""
    phi = -np.arange(M + 1) * DOS_PI / M
    u1, u2 = math.cos(th), math.sin(th)
    P = np.stack([r * np.cos(phi), r * np.sin(phi) * u1, Z0 + r * np.sin(phi) * u2], axis=1)
    return (P[1:] + P[:-1]) / 2, P[1:] - P[:-1]


# Campo magnético de la bobina sumando el aporte de cada segmento (ley de Biot-Savart).
def b_lazo(mid, dl, obs, N: int, I: float, eps2: float = 1e-6):
    """Biot-Savart numérico: B = (μ0 N I / 4π) Σ dl × R / |R|³  para puntos `obs` (G,3)."""
    R = obs[:, None, :] - mid[None, :, :]
    r2 = (R ** 2).sum(-1) + eps2
    c = np.cross(dl[None, :, :], R)
    return MU0 * N * I / (4 * PI) * (c / (r2 * np.sqrt(r2))[..., None]).sum(1)


# Valores finales (velocidad estable, calor, rendimiento) con una fórmula rápida, sin simular. Se usa en la tabla.
def escenario(p: Params, N: int, V: float, T: float = T_AMB) -> dict:
    """Régimen estacionario en forma cerrada (giro rápido) para N espiras y V voltios."""
    q = replace(p, N=int(N), V=float(V))
    g = geometria(q)
    Rb = g['R0'] * (1 + ALFA_T * (T - T_AMB))
    R = Rb + q.Rext
    K0, I0 = g['K0'], q.V / R
    wss = (K0 * q.V / (PI * R)) / (q.b + K0 ** 2 / (4 * R)) if K0 > 0 else 0.0
    w0 = 4 * q.V / (PI * K0) if K0 > 0 else 0.0
    Imed = (PI * q.V - 2 * K0 * wss) / (2 * PI * R)
    I2 = (PI * q.V ** 2 - 4 * q.V * K0 * wss + 0.5 * PI * K0 ** 2 * wss ** 2) / (2 * PI * R ** 2)
    Pin, Pj, Pext, Pf = q.V * Imed, Rb * I2, q.Rext * I2, q.b * wss ** 2
    return dict(N=int(N), V=float(V), Rb=Rb, R=R, I0=I0, tauMax=K0 * I0, wss=wss, w0=w0,
                Imed=Imed, Pin=Pin, Pj=Pj, Pext=Pext, Pf=Pf,
                eta=Pf / Pin if Pin > 0 else 0.0, Tss=T_AMB + Pj / (H_CONV * g['S']),
                gen=wss > q.V / max(K0, 1e-12), K0=K0, g=g)


print("Celda 1 lista: modelo físico cargado.")
```

Se realizan gráficas del motor en 3D, el mapa del campo magnético y las curvas de corriente, torque, velocidad y temperatura.

```python
# GRÁFICAS: escena 3D, mapa del campo, series temporales
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec
from matplotlib.patches import Circle, Rectangle
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401


# Toma de la simulación el estado del motor en un instante.
def estado_en(sim: dict, k: int) -> dict:
    """Estado del motor en el índice k de la simulación."""
    return dict(t=sim['t'][k], th=sim['th'][k], om=sim['om'][k], T=sim['T'][k],
                I=sim['I'][k], tau=sim['tau'][k], eps=sim['eps'][k])


# Dibuja el motor en 3D: imán, bobina y las flechas de campo, momento, torque, fuerzas y corriente.
def escena3d(ax, p: Params, g: dict, st: dict, mostrar=("B", "mu", "tau", "F", "E")) -> None:
    """Dibuja el imán, la bobina, los soportes y los vectores B, μ, τ, dF y la corriente."""
    ax.cla()
    th, I, r = st['th'], st['I'], p.r
    on, hayI = conmutador(th), abs(st['I']) > 1e-9
    sg = 1.0 if I >= 0 else -1.0
    n = np.array([0.0, math.sin(th), -math.cos(th)])
    u = np.array([0.0, math.cos(th), math.sin(th)])
    c0 = np.array([0.0, 0.0, Z0])

    # imán y su base
    ang = np.linspace(0, DOS_PI, 64)
    ax.plot_surface(R_IMAN * np.outer(np.cos(ang), [1, 1]), R_IMAN * np.outer(np.sin(ang), [1, 1]),
                    np.outer(np.ones(64), [-0.010, 0.0]), color='#7f8c8d', alpha=0.55, shade=False, linewidth=0)
    TH_, RR_ = np.meshgrid(ang, [0.0, R_IMAN])
    ax.plot_surface(RR_ * np.cos(TH_), RR_ * np.sin(TH_), np.zeros_like(TH_), color='#c0392b',
                    alpha=0.5, shade=False, linewidth=0)

    # soportes, ejes, alambre de la mitad pelada y el contacto
    for s in (-1, 1):
        ax.plot([s * X_SOP] * 2, [0, 0], [0, Z0 - 0.002], color='dimgray', lw=5, solid_capstyle='round')
        ax.plot([s * X_SOP, s * r], [0, 0], [Z0, Z0], color='#c97b3a', lw=2.5)
    ax.plot([X_SOP - 0.008] * 2, [0, -0.011 * math.cos(th)], [Z0, Z0 - 0.011 * math.sin(th)],
            color='gold', lw=4)
    ax.plot([X_SOP], [0], [Z0], 'o', color='limegreen' if on else 'gray', ms=9, mec='k')

    # espiras (máximo 6 dibujadas)
    nd = max(1, min(6, int(p.N)))
    phi = np.linspace(0, DOS_PI, 73)
    for kk in range(nd):
        off = (kk - (nd - 1) / 2) * 0.0013
        ax.plot(r * np.cos(phi) + off * n[0], r * np.sin(phi) * u[1] + off * n[1],
                Z0 + r * np.sin(phi) * u[2] + off * n[2],
                color='#ffa14a' if (hayI and on) else '#c97b3a', lw=2.0)

    if "B" in mostrar:
        ax.quiver(0, -0.04, 0, 0, 0, 0.045, color='tab:blue', lw=3, arrow_length_ratio=0.22)
        ax.text(0, -0.04, 0.050, "B imán", color='tab:blue', fontsize=9, ha='center')

    if hayI:
        # corriente / campo E: tangente en el sentido de I (φ decreciente)
        if "E" in mostrar:
            ph = -(np.arange(10) + 0.5) * DOS_PI / 10
            P = np.stack([r * np.cos(ph), r * np.sin(ph) * u[1], Z0 + r * np.sin(ph) * u[2]], 1)
            Tn = np.stack([sg * r * np.sin(ph), -sg * r * np.cos(ph) * u[1], -sg * r * np.cos(ph) * u[2]], 1)
            Tn /= np.linalg.norm(Tn, axis=1)[:, None]
            ax.quiver(P[:, 0], P[:, 1], P[:, 2], Tn[:, 0] * 0.0075, Tn[:, 1] * 0.0075, Tn[:, 2] * 0.0075,
                      color='#e6c200', lw=1.6, arrow_length_ratio=0.6, normalize=False)
        # fuerzas dF = N I dl × B  (B = (0,0,B) ⇒ dF ∝ (dl_y, −dl_x, 0): siempre horizontal)
        if "F" in mostrar:
            ph = -(np.arange(16) + 0.5) * DOS_PI / 16
            P = np.stack([r * np.cos(ph), r * np.sin(ph) * u[1], Z0 + r * np.sin(ph) * u[2]], 1)
            tx, ty = r * np.sin(ph), -r * np.cos(ph) * u[1]
            fx, fy = sg * ty / r, -sg * tx / r
            ok = np.hypot(fx, fy) > 0.04
            ax.quiver(P[ok, 0], P[ok, 1], P[ok, 2], fx[ok] * 0.016, fy[ok] * 0.016, np.zeros(ok.sum()),
                      color='#ff9f1c', lw=1.6, arrow_length_ratio=0.35, normalize=False)
        if "mu" in mostrar:
            v = sg * n * 0.04
            ax.quiver(*c0, *v, color='tab:red', lw=3, arrow_length_ratio=0.25)
            ax.text(*(c0 + v * 1.25), "μ", color='tab:red', fontsize=11, fontweight='bold')
        if "tau" in mostrar:
            Iref = p.V / (g['R0'] * (1 + ALFA_T * (st['T'] - T_AMB)) + p.Rext)
            f = np.clip(st['tau'] / max(1e-12, g['K0'] * abs(Iref)), -1, 1)
            if abs(f) > 0.03:
                ax.quiver(*c0, f * 0.05, 0, 0, color='tab:green', lw=3.4, arrow_length_ratio=0.25)
                ax.text(*(c0 + np.array([f * 0.058, 0, 0.004])), "τ", color='tab:green', fontsize=11,
                        fontweight='bold')
    else:
        ax.plot([0, n[0] * 0.035], [0, n[1] * 0.035], [Z0, Z0 + n[2] * 0.035], ls='--', color='#ff8080', lw=1.2)
        ax.text(0, 0, Z0 + 0.04, "circuito abierto: I = 0 ⇒ μ = 0, τ = 0", fontsize=8, ha='center')

    ax.set_xlim(-0.07, 0.07); ax.set_ylim(-0.06, 0.06); ax.set_zlim(-0.012, 0.075)
    try:
        ax.set_box_aspect((0.14, 0.12, 0.087))
    except AttributeError:
        pass
    ax.view_init(elev=24, azim=-58)
    ax.set_xlabel("X (eje)", fontsize=8); ax.set_ylabel("Y", fontsize=8); ax.set_zlabel("Z (B ↑)", fontsize=8)
    ax.tick_params(labelsize=6)
    ax.set_title(f"θ = {math.degrees(th % DOS_PI):.0f}°   ω = {st['om']:.1f} rad/s   I = {I:.2f} A   "
                 f"({'contacto' if on else 'abierto'})", fontsize=9)


# Mapa del campo magnético en un corte plano
def mapa2d(ax, p: Params, g: dict, st: dict, modo: str = "total") -> None:
    """Campo magnético en el plano YZ (x = 0): imán uniforme + bobina por Biot-Savart."""
    ax.cla()
    y0, y1, za, zb = -0.058, 0.058, -0.014, 0.066
    NX, NZ = 29, 19
    ys = y0 + (np.arange(NX) + 0.5) * (y1 - y0) / NX
    zs = za + (np.arange(NZ) + 0.5) * (zb - za) / NZ
    YY, ZZ = np.meshgrid(ys, zs, indexing='ij')
    obs = np.stack([np.zeros(YY.size), YY.ravel(), ZZ.ravel()], axis=1)
    Rb = g['R0'] * (1 + ALFA_T * (st['T'] - T_AMB))
    Iplot = st['I'] if abs(st['I']) > 1e-9 else p.V / (Rb + p.Rext)
    mid, dl = lazo(st['th'], p.r)
    By = np.zeros(len(obs)); Bz = np.zeros(len(obs))
    if modo != "bobina":
        Bz += p.B
    if modo != "iman":
        bb = b_lazo(mid, dl, obs, p.N, Iplot)
        By += bb[:, 1]; Bz += bb[:, 2]
    M = np.hypot(By, Bz)
    lm = np.log10(np.maximum(M, 1e-12))
    lo, hi = np.percentile(lm, 4), np.percentile(lm, 96)
    q = ax.quiver(obs[:, 1], obs[:, 2], By / np.maximum(M, 1e-30), Bz / np.maximum(M, 1e-30), lm,
                  cmap='turbo', clim=(lo, hi), pivot='mid', scale=34, width=0.0028)
    ax.add_patch(Rectangle((-R_IMAN, -0.010), 2 * R_IMAN, 0.010, color='#7f8c8d', alpha=0.9, zorder=3))
    ax.add_patch(Rectangle((-R_IMAN, -0.0015), 2 * R_IMAN, 0.0015, color='#c0392b', zorder=4))
    sg = 1 if Iplot >= 0 else -1
    for s in (1, -1):
        cy, cz = s * p.r * math.cos(st['th']), Z0 + s * p.r * math.sin(st['th'])
        ax.add_patch(Circle((cy, cz), 0.0035, fc='#10151f', ec='#ffa14a', lw=2, zorder=6))
        ax.text(cy, cz, "⊙" if s * sg > 0 else "⊗", color='#ffe14d', ha='center', va='center',
                fontsize=11, zorder=7)
    bc = MU0 * p.N * Iplot / (2 * p.r)
    ax.set_title(f"Campo B en el plano YZ ({'total' if modo == 'total' else 'solo ' + modo}) · "
                 f"B bobina(centro) = {abs(bc) * 1e3:.2f} mT = {abs(bc) / p.B * 100:.1f} % del imán", fontsize=9)
    ax.set_xlim(y0, y1); ax.set_ylim(za, zb); ax.set_aspect('equal')
    ax.set_xlabel("y [m]", fontsize=8); ax.set_ylabel("z [m]", fontsize=8); ax.tick_params(labelsize=7)
    return q


# 4 gráficas: corriente, torque, velocidad y temperatura.
def series(axs, p: Params, g: dict, sim: dict, k: int) -> None:
    """Cuatro gráficas: I(t) y τ(t) con zoom alrededor del instante k; ω(t) y T(t) completas."""
    t = sim['t']
    ax_i, ax_tau, ax_w, ax_T = axs
    tv = t[k]
    om_v = max(abs(sim['om'][k]), 2.0)
    mitad = min(3 * DOS_PI / om_v, (t[-1] - t[0]) / 2 + 1e-9)
    a, b = max(t[0], tv - mitad), min(t[-1], tv + mitad)
    if b - a < 2 * mitad - 1e-9:                      # mantener el ancho de la ventana en los bordes
        a, b = (t[0], t[0] + 2 * mitad) if a <= t[0] else (t[-1] - 2 * mitad, t[-1])
    m = (t >= a) & (t <= b)
    R0 = g['R0']; I0 = p.V / (R0 + p.Rext)

    ax_i.plot(t[m], sim['I'][m], color='#e0524e', lw=1.4)
    ax_i.axhline(I0, color='gray', ls='--', lw=0.9)
    ax_i.text(a, I0, "  I₀ = V/R (rotor frenado)", fontsize=7, va='bottom', color='gray')
    ax_i.set_title("Corriente I(t) (zoom ~6 vueltas)", fontsize=9); ax_i.set_ylabel("A", fontsize=8)

    ax_tau.plot(t[m], sim['tau'][m], color='#2fa866', lw=1.4)
    ax_tau.plot(t[m], p.b * sim['om'][m], color='gray', ls='--', lw=0.9, label="b·ω (fricción)")
    ax_tau.legend(fontsize=7, loc='upper right')
    ax_tau.set_title("Torque magnético τ(t)", fontsize=9); ax_tau.set_ylabel("N·m", fontsize=8)

    ss = escenario(p, p.N, p.V, float(sim['T'][-1]))
    ax_w.plot(t, sim['om'], color='#9b6fd0', lw=1.2)
    ax_w.axhline(ss['wss'], color='k', ls='--', lw=1.0, label=f"ω_ss teórica = {ss['wss']:.1f} rad/s")
    ax_w.legend(fontsize=7, loc='lower right')
    ax_w.set_title("Velocidad angular ω(t) (toda la simulación)", fontsize=9); ax_w.set_ylabel("rad/s", fontsize=8)

    ax_T.plot(t, sim['T'], color='#e08a3c', lw=1.4)
    ax_T.axhline(T_ESMALTE, color='red', ls='--', lw=0.9, label=f"límite del esmalte ({T_ESMALTE:.0f} °C)")
    ax_T.legend(fontsize=7, loc='upper left')
    ax_T.set_title("Temperatura del alambre", fontsize=9); ax_T.set_ylabel("°C", fontsize=8)

    for ax_ in (ax_i, ax_tau):
        ax_.axvline(tv, color='tab:blue', lw=1.0, alpha=0.8)
        ax_.set_xlim(a, b)
    for ax_ in (ax_w, ax_T):
        ax_.axvline(tv, color='tab:blue', lw=1.0, alpha=0.8)
        ax_.set_xlim(t[0], t[-1])
    for ax_ in axs:
        ax_.grid(alpha=0.3); ax_.tick_params(labelsize=7); ax_.set_xlabel("t [s]", fontsize=8)


# Juntar todo en una sola figura
def construir_figura(p: Params, sim: dict, k: int, modo_mapa: str = "total", mostrar=("B", "mu", "tau", "F", "E")):
    """Figura completa: 3D + mapa de campo + 4 series temporales para el instante k."""
    g = sim['g']
    fig = plt.figure(figsize=(15, 11), dpi=80)
    gs = GridSpec(3, 2, figure=fig, height_ratios=[1.7, 0.85, 0.85], hspace=0.42, wspace=0.18,
                  left=0.05, right=0.98, top=0.91, bottom=0.05)
    ax3 = fig.add_subplot(gs[0, 0], projection='3d')
    axm = fig.add_subplot(gs[0, 1])
    axs = [fig.add_subplot(gs[1, 0]), fig.add_subplot(gs[2, 0]), fig.add_subplot(gs[1, 1]), fig.add_subplot(gs[2, 1])]
    st = estado_en(sim, k)
    escena3d(ax3, p, g, st, mostrar)
    q = mapa2d(axm, p, g, st, modo_mapa)
    cb = fig.colorbar(q, ax=axm, fraction=0.035, pad=0.01)
    cb.set_label("log₁₀ |B| [T]", fontsize=8); cb.ax.tick_params(labelsize=7)
    series(axs, p, g, sim, k)
    fig.suptitle(f"Motor casero · N = {p.N} espiras · V = {p.V:g} V · B = {p.B * 1e3:g} mT · t = {st['t']:.3f} s",
                 fontsize=12, fontweight='bold')
    return fig


print("Celda 2 lista: funciones de gráficas cargadas.")
```

### Procedimiento y cálculos 
Prepara los textos con las fórmulas y una tabla que recopila las variaciones cuando las condiciones son diferentes a las inciciales.

```python
# Formato a los números y unidades
def fm(x: float, u: str = "", dig: int = 3) -> str:
    """Formato compacto con notación científica """
    if not np.isfinite(x):
        return "—"
    if x == 0:
        s = "0"
    elif abs(x) >= 1e4 or abs(x) < 1e-2:
        m, e = f"{x:.{dig - 1}e}".split("e")
        s = f"{m}×10<sup>{int(e)}</sup>"
    else:
        s = f"{float(f'{x:.{dig}g}'):g}"
    return s + (f" {u}" if u else "")


def eq(simb: str, formula: str, sust: str, val: float, unidad: str = "") -> str:
    partes = f"<b style='color:#1a4fb3'>{simb}</b>" + (f" = {formula}" if formula else "")
    if sust:
        partes += f" = {sust}"
    return f"<div style='padding:2px 0;border-bottom:1px dotted #ccc'>{partes} = <b>{fm(val, unidad)}</b></div>"


def nota(txt: str) -> str:
    return f"<div style='color:#666;font-style:italic;font-size:90%;margin:3px 0'>{txt}</div>"


def linea(txt: str) -> str:
    return f"<div style='padding:2px 0;border-bottom:1px dotted #ccc'>{txt}</div>"


# Realiza los cálculos del instante representativo θ = 90° y devuelve un bloque de texto con las fórmulas y resultados.
def bloques_calculo(p: Params, sim: dict) -> list:
    """Secciones con los cálculos del instante representativo """
    g = sim['g']
    n = len(sim['t'])
    omR = float(np.mean(sim['om'][int(0.8 * n):]))             # ω medio al final de la simulación
    Tn = float(sim['T'][-1])
    Rb = g['R0'] * (1 + ALFA_T * (Tn - T_AMB)); R = Rb + p.Rext; I0 = p.V / R
    K0 = g['K0']; eps = K0 * omR; I = (p.V - eps) / R; tau = K0 * I
    Ic = I
    bC = MU0 * p.N * Ic / (2 * p.r)
    btot = math.hypot(bC * 1.0, p.B + 0.0)                      # θ = 90°: n̂ = (0, 1, 0) ⊥ ẑ
    Jd = I / g['Aw']; rhoT = RHO * (1 + ALFA_T * (Tn - T_AMB)); Ec = rhoT * Jd; vd = Jd / (N_E * Q_E)
    sumF = p.N * abs(I) * p.B * 4 * p.r
    ss = escenario(p, p.N, p.V, Tn)
    wss, w0 = ss['wss'], ss['w0']
    tm = g['J'] / (p.b + K0 ** 2 / (4 * R))
    Iprom = float(np.mean(sim['I'][int(0.8 * n):]))
    Pin = p.V * Iprom
    I2 = float(np.mean(sim['I'][int(0.8 * n):] ** 2))
    Pj, Pe = Rb * I2, p.Rext * I2
    Pem = float(np.mean(sim['tau'][int(0.8 * n):] * sim['om'][int(0.8 * n):]))
    Pf = p.b * float(np.mean(sim['om'][int(0.8 * n):] ** 2))
    dTdt0 = I0 ** 2 * Rb / g['Cth']
    bz = [MU0 * p.N * Ic * p.r ** 2 / (2 * (p.r ** 2 + z ** 2) ** 1.5) for z in (0.02, 0.05)]

    s1 = "".join([
        eq("A", "π r²", f"π({fm(p.r, 'm')})²", g['A'], "m²"),
        eq("ℓ", "N · 2πr", f"{p.N} · 2π({fm(p.r, 'm')})", g['l'], "m"),
        eq("A<sub>w</sub>", "π d² / 4", f"π({fm(p.d * 1e3, 'mm')})² / 4", g['Aw'], "m²"),
        eq("R<sub>bob,20°C</sub>", "ρ ℓ / A<sub>w</sub>", f"{fm(RHO, 'Ω·m')} · {fm(g['l'], 'm')} / {fm(g['Aw'], 'm²')}", g['R0'], "Ω"),
        eq("R<sub>bob</sub>(T)", "R<sub>bob,20°C</sub> [1 + α(T − 20)]", f"{fm(g['R0'], 'Ω')} [1 + {ALFA_T}({fm(Tn)} − 20)]", Rb, "Ω"),
        eq("R", "R<sub>bob</sub> + R<sub>ext</sub>", f"{fm(Rb, 'Ω')} + {fm(p.Rext, 'Ω')}", R, "Ω"),
        eq("m", "ρ<sub>Cu</sub> ℓ A<sub>w</sub>", f"{fm(DENS, 'kg/m³')} · {fm(g['l'], 'm')} · {fm(g['Aw'], 'm²')}", g['m'], "kg"),
        eq("J", "J<sub>eje</sub> + ½ m r²", f"{fm(J_AXLE, 'kg·m²')} + ½ · {fm(g['m'], 'kg')} · ({fm(p.r, 'm')})²", g['J'], "kg·m²"),
        nota("Más espiras ⇒ más longitud de alambre ⇒ más resistencia (R ∝ N), más masa y más inercia."),
    ])
    s2 = "".join([
        nota(f"Instante representativo: θ = 90° (contacto cerrado, torque máximo) con ω = {fm(omR, 'rad/s')} "
             "(promedio del final de la simulación)."),
        eq("I<sub>0</sub>", "V / R", f"{fm(p.V, 'V')} / {fm(R, 'Ω')}", I0, "A"),
        nota("Corriente con el rotor frenado (sin fem inducida): la mayor posible."),
        eq("ε", "N A B ω sinθ", f"{p.N} · {fm(g['A'], 'm²')} · {fm(p.B, 'T')} · {fm(omR, 'rad/s')} · sin 90°", eps, "V"),
        nota("Ley de Faraday: Φ = −N A B cosθ ⇒ ε = |dΦ/dt| = N A B ω sinθ; se opone a V (ley de Lenz)."),
        eq("ε<sub>fuente</sub> (fem de la fuente)", "V", "", p.V, "V"),
        eq("K (constante de la fem)", "N A B", f"{p.N} · {fm(g['A'], 'm²')} · {fm(p.B, 'T')}", K0, "V·s"),
        eq("ε<sub>máx</sub> (fem inducida máxima)", "N A B ω", f"{fm(K0, 'V·s')} · {fm(omR, 'rad/s')}", K0 * omR, "V"),
        eq("ε<sub>prom</sub> (mientras hay contacto)", "(2/π) N A B ω", "", 2 * K0 * omR / PI, "V"),
        eq("fem neta", "V − ε", f"{fm(p.V, 'V')} − {fm(eps, 'V')}", p.V - eps, "V"),
        eq("ε<sub>máx</sub> / V", "100 · ε<sub>máx</sub> / V", "", K0 * omR / p.V * 100 if p.V > 0 else 0.0, "%"),
        linea("<b>I</b> = (V − ε)/R  si 0 ≤ θ mod 2π &lt; π ;   <b>I</b> = 0  si π ≤ θ mod 2π &lt; 2π"),
        eq("I (θ = 90°)", "(V − ε) / R", f"({fm(p.V, 'V')} − {fm(eps, 'V')}) / {fm(R, 'Ω')}", I, "A"),
        eq("⟨I⟩ (promedio de la simulación)", "", "", Iprom, "A"),
    ])
    s3 = "".join([
        linea(f"<b>B<sub>imán</sub></b> = (0, 0, B) = <b>{fm(p.B * 1e3, 'mT')}</b> hacia arriba (uniforme sobre la bobina)"),
        linea("<b>Biot-Savart</b>: B(P) = (μ₀ N I / 4π) Σ dl × R / |R|³  (48 segmentos, usado en el mapa)"),
        eq("B<sub>bobina</sub>(centro)", "μ₀ N I / (2r)", f"{fm(MU0, 'T·m/A')} · {p.N} · {fm(Ic, 'A')} / (2 · {fm(p.r, 'm')})", bC, "T"),
        eq("B<sub>bobina</sub>(z = 2 cm)", "μ₀ N I r² / [2(r²+z²)<sup>3/2</sup>]", "", bz[0], "T"),
        eq("B<sub>bobina</sub>(z = 5 cm)", "— ídem —", "", bz[1], "T"),
        eq("|B<sub>total</sub>|(centro)", "| B<sub>imán</sub> ẑ + B<sub>bobina</sub> n̂ |", "n̂ = (0, 1, 0) a θ = 90°", btot, "T"),
        eq("B<sub>bobina</sub> / B<sub>imán</sub>", "", "", bC / p.B * 100, "%"),
        nota("El campo propio de la bobina es pequeño frente al del imán con pocas espiras; con muchas (por ejemplo 100) "
             "llega a ser comparable y deforma el mapa."),
    ])
    s4 = "".join([
        eq("J", "I / A<sub>w</sub>", f"{fm(I, 'A')} / {fm(g['Aw'], 'm²')}", Jd, "A/m²"),
        eq("E", "ρ(T) J", f"{fm(rhoT, 'Ω·m')} · {fm(Jd, 'A/m²')}", Ec, "V/m"),
        linea(f"Comprobación: E · ℓ = {fm(Ec * g['l'], 'V')} = I · R<sub>bob</sub> = {fm(I * Rb, 'V')}"),
        eq("v<sub>d</sub>", "J / (n e)", f"n = {fm(N_E, 'm⁻³')}", vd, "m/s"),
        eq("E<sub>mov</sub> máx", "|v × B| = ω r B |sinθ|", f"{fm(omR, 'rad/s')} · {fm(p.r, 'm')} · {fm(p.B, 'T')}", abs(omR * p.r * p.B), "V/m"),
        linea("<b>Kirchhoff</b>: V = I R<sub>bob</sub> + I R<sub>ext</sub> + ε"),
        linea(f"{fm(p.V, 'V')} = {fm(I * Rb, 'V')} + {fm(I * p.Rext, 'V')} + {fm(eps, 'V')} = <b>{fm(I * Rb + I * p.Rext + eps, 'V')}</b>"),
    ])
    s5 = "".join([
        linea("<b>Fuerza de Lorentz</b>: dF = N I dl × B ; con B = (0,0,B): dF = N I B (dl<sub>y</sub>, −dl<sub>x</sub>, 0), siempre horizontal"),
        eq("|dF| máx por metro", "N I B", f"{p.N} · {fm(abs(I), 'A')} · {fm(p.B, 'T')}", p.N * abs(I) * p.B, "N/m"),
        eq("Σ|dF| (θ = 90°)", "N I B · 4r", "", sumF, "N"),
        linea("<b>Fuerza neta</b>: F = N I (∮dl) × B = <b>0</b> (lazo cerrado, B uniforme): la bobina no se traslada, solo gira."),
        eq("μ", "N I A", f"{p.N} · {fm(I, 'A')} · {fm(g['A'], 'm²')}", p.N * I * g['A'], "A·m²"),
        linea("μ = μ n̂ = (0, μ sinθ, −μ cosθ) ;  τ = μ × B = (μ<sub>y</sub> B, 0, 0)"),
        eq("τ<sub>x</sub>", "N I A B sinθ", f"{p.N} · {fm(I, 'A')} · {fm(g['A'], 'm²')} · {fm(p.B, 'T')} · 1", tau, "N·m"),
        eq("τ<sub>máx</sub> (rotor frenado)", "N I₀ A B", "", K0 * I0, "N·m"),
        eq("τ<sub>fricción</sub>", "b ω", f"{fm(p.b, 'N·m·s')} · {fm(omR, 'rad/s')}", p.b * omR, "N·m"),
        eq("α", "(τ − b ω)/J", f"({fm(tau, 'N·m')} − {fm(p.b * omR, 'N·m')}) / {fm(g['J'], 'kg·m²')}", (tau - p.b * omR) / g['J'], "rad/s²"),
        eq("Peso del cobre", "m g", "", g['m'] * G, "N"),
        eq("Reacción en cada clip", "m g / 2", "", g['m'] * G / 2, "N"),
        eq("Atracción entre espiras vecinas (por metro)", "μ₀ I² / (2π s)", f"s ≈ d = {fm(p.d * 1e3, 'mm')}", MU0 * I * I / (2 * PI * p.d), "N/m"),
        eq("Fuerza magnética sobre un electrón", "e v<sub>d</sub> B", "", Q_E * vd * p.B, "N"),
    ])
    s6 = "".join([
        linea(f"Durante media vuelta: τ = K sinθ (V − K ω sinθ)/R, con K = N A B = {fm(K0, 'V·s')}"),
        linea("⟨τ⟩ = (1/2π) ∫<sub>0</sub><sup>π</sup> K sinθ (V − K ω sinθ)/R dθ = K V/(πR) − K² ω/(4R)"),
        linea("[∫<sub>0</sub><sup>π</sup> sinθ dθ = 2 ;  ∫<sub>0</sub><sup>π</sup> sin²θ dθ = π/2]"),
        linea("En régimen estacionario ⟨τ⟩ = b ω ⇒ <b>ω<sub>ss</sub> = (K V/πR) / (b + K²/4R)</b>"),
        eq("ω<sub>ss</sub>", "(K V/πR) / (b + K²/4R)", f"({fm(K0)} · {fm(p.V)} / (π · {fm(R)})) / ({fm(p.b)} + {fm(K0 ** 2 / (4 * R))})", wss, "rad/s"),
        linea(f"= <b>{fm(wss * 60 / DOS_PI, 'rpm')}</b> ;  simulación (final): {fm(omR, 'rad/s')} = {fm(omR * 60 / DOS_PI, 'rpm')}"),
        eq("ω<sub>0</sub> sin fricción", "4V / (π K)", "", w0, "rad/s"),
        eq("Constante de tiempo mecánica", "J / (b + K²/4R)", "", tm, "s"),
        nota("Aquí ω_ss supera V/K: en parte del ciclo ε &gt; V (I &lt; 0); la fórmula es aproximada." if ss['gen']
             else "La fórmula supone giro rápido; con pulsos marcados la simulación puede diferir unos %."),
    ])
    s7 = "".join([
        eq("P<sub>fuente</sub>", "V ⟨I⟩", f"{fm(p.V, 'V')} · {fm(Iprom, 'A')}", Pin, "W"),
        eq("P<sub>Joule,bobina</sub>", "⟨I² R<sub>bob</sub>⟩", "", Pj, "W"),
        eq("P<sub>Joule,externa</sub>", "⟨I² R<sub>ext</sub>⟩", "", Pe, "W"),
        eq("P<sub>electromagnética</sub>", "⟨τ ω⟩ = ⟨I ε⟩", "", Pem, "W"),
        eq("P<sub>fricción</sub>", "b ⟨ω²⟩", "", Pf, "W"),
        linea(f"<b>Balance</b>: {fm(Pin, 'W')} ≈ {fm(Pj + Pe, 'W')} + {fm(Pem, 'W')} = <b>{fm(Pj + Pe + Pem, 'W')}</b>"),
        eq("Rendimiento η", "P<sub>fricción</sub> / P<sub>fuente</sub>", "(toda la energía mecánica se va en fricción)", Pf / Pin * 100 if Pin > 1e-9 else 0.0, "%"),
        nota("El rendimiento es bajo porque el conmutador abre el circuito media vuelta y mucha energía se disipa en las resistencias."),
    ])
    s8 = "".join([
        eq("C<sub>térmica</sub>", "m c<sub>Cu</sub>", "", g['Cth'], "J/K"),
        eq("Superficie de enfriamiento", "ℓ π d", "", g['S'], "m²"),
        linea(f"dT/dt = [I² R<sub>bob</sub> − h S (T − T<sub>amb</sub>)] / (m c), con h = {H_CONV:g} W/m²K"),
        eq("dT/dt inicial (rotor frenado)", "I₀² R<sub>bob</sub> / (m c)", "", dTdt0, "°C/s"),
        eq("T<sub>ss</sub> estimada", "T<sub>amb</sub> + ⟨P<sub>Joule</sub>⟩ / (h S)", "", ss['Tss'], "°C"),
        nota(f"Si T<sub>ss</sub> supera {T_ESMALTE:.0f} °C el esmalte se degrada. El calor crece como I²: duplicar V multiplica ≈ ×4 el calor."),
    ])
    return [("1 · Geometría, masa y resistencia", s1), ("2 · Corriente (Ohm con fem inducida)", s2),
            ("3 · Campo magnético (imán, bobina, total)", s3), ("4 · Campo eléctrico", s4),
            ("5 · Fuerzas y torque", s5), ("6 · Velocidad terminal (deducción)", s6),
            ("7 · Balance de energía", s7), ("8 · Calentamiento", s8)]


# Tabla que compara el motor con el doble/mitad de espiras y de voltaje.
def tabla_escenarios(p: Params, T: float = T_AMB) -> str:
    """Tabla en régimen estacionario: tu configuración vs ×2 espiras, ½ espiras, ×2 V, ×2 y ×2."""
    esc = [("Configuración actual", p.N, p.V), ("× 2 espiras", p.N * 2, p.V),
           ("½ espiras", max(1, round(p.N / 2)), p.V), ("× 2 voltaje", p.N, p.V * 2),
           ("× 2 espiras y × 2 voltaje", p.N * 2, p.V * 2)]
    cab = ["Escenario", "N", "V (V)", "R (Ω)", "I₀ (A)", "τ máx (N·m)", "ω_ss (rad/s)", "rpm", "P Joule (W)", "η (%)", "T_ss (°C)"]
    h = "<table style='border-collapse:collapse;font-size:90%'><tr>" + "".join(
        f"<th style='padding:4px 8px;border-bottom:2px solid #888;text-align:right'>{c}</th>" for c in cab) + "</tr>"
    for i, (nom, N, V) in enumerate(esc):
        r = escenario(p, N, V, T)
        fondo = "background:#e8f0ff;" if i == 0 else ""
        rojo = "color:#c0392b;font-weight:bold;" if r['Tss'] > T_ESMALTE else ""
        celdas = [nom + (" (generador)" if r['gen'] else ""), fm(r['N']), fm(r['V']), fm(r['R']), fm(r['I0']), fm(r['tauMax']),
                  fm(r['wss']), fm(r['wss'] * 60 / DOS_PI), fm(r['Pj'] + r['Pext']), fm(r['eta'] * 100, "", 2), fm(r['Tss'])]
        h += f"<tr style='{fondo}'>" + "".join(
            f"<td style='padding:3px 8px;border-bottom:1px solid #ddd;text-align:{'left' if j == 0 else 'right'};{rojo if j == 10 else ''}'>{c}</td>"
            for j, c in enumerate(celdas)) + "</tr>"
    return h + "</table>"


# Advertencias: corriente muy alta, temperatura peligrosa, motor detenido, etc.
def alertas_html(p: Params, sim: dict) -> str:
    """Mensajes sobre lo que pasa con los valores elegidos."""
    g = sim['g']
    Rb = g['R0'] * (1 + ALFA_T * (float(sim['T'][-1]) - T_AMB)); R = Rb + p.Rext; I0 = p.V / R
    Tmax = float(np.max(sim['T']))
    n = len(sim['t']); fin = slice(int(0.8 * n), n)
    msgs = []
    if p.V == 0:
        msgs.append(("#a8741a", "Voltaje 0 V: no circula corriente, no hay torque."))
    if I0 > 8:
        msgs.append(("#b22222", f"Corriente de arranque {fm(I0, 'A')}: una pila real no la entrega (su resistencia interna es mayor) y el alambre se calentaría muy rápido."))
    elif I0 > 3:
        msgs.append(("#a8741a", f"Corriente alta ({fm(I0, 'A')}): el alambre y la pila se calentarán en pocos segundos."))
    if Tmax > T_ESMALTE + 25:
        msgs.append(("#b22222", f"¡Esmalte dañado! La temperatura llegó a {Tmax:.0f} °C (límite {T_ESMALTE:.0f} °C): las espiras se pondrían en cortocircuito."))
    elif Tmax > T_ESMALTE:
        msgs.append(("#b22222", f"Temperatura {Tmax:.0f} °C sobre el límite del esmalte ({T_ESMALTE:.0f} °C)."))
    elif Tmax > 80:
        msgs.append(("#a8741a", f"El alambre llega a {Tmax:.0f} °C y sigue subiendo."))
    if p.V > 0 and abs(sim['om'][-1]) < 0.5 and not conmutador(float(sim['th'][-1])):
        msgs.append(("#a8741a", "La bobina se detuvo en la media vuelta sin corriente: sube la «Velocidad inicial (impulso)» para arrancarla, como en el experimento real."))
    if sim['I'][fin].min() < -0.05 * max(abs(sim['I'][fin]).max(), 1e-9):
        msgs.append(("#1f5fae", "En parte del ciclo I < 0: la fem inducida supera a V y el motor actúa como generador."))
    if not msgs:
        msgs.append(("#1c7a3d", "Funcionamiento normal. Prueba duplicar las espiras o el voltaje y mira la tabla de escenarios."))
    return "".join(f"<span style='display:inline-block;margin:2px 6px 2px 0;padding:4px 10px;border-radius:10px;"
                   f"border:1px solid {c};color:{c};font-size:90%'>{t}</span>" for c, t in msgs)


print("Celda 3 lista: cálculos, alertas y tabla cargados.")
```

```python

import ipywidgets as W
from IPython.display import display, clear_output, HTML
from matplotlib.animation import FuncAnimation


# Construye el panel: cajas y deslizadores, botones de ejemplo, gráficas y cálculos.
def lanzar():
    """Construye el panel interactivo y lo muestra."""
    est = dict(sim=None, p=None, bloqueo=False)
    ancho = W.Layout(width='320px')
    estilo = {'description_width': '95px'}

    # Entradas principales: espiras y voltaje
    t_N = W.IntText(value=6, description='Espiras N', style=estilo, layout=W.Layout(width='190px'))
    s_N = W.IntSlider(value=6, min=1, max=200, continuous_update=False, readout=False, layout=ancho)
    t_V = W.FloatText(value=9.0, description='Voltaje V', style=estilo, layout=W.Layout(width='190px'))
    s_V = W.FloatSlider(value=9.0, min=0, max=24, step=0.1, continuous_update=False, readout=False, layout=ancho)

    # Otros parámetros: campo, fricción, alambre, radio, resistencia externa, duración, condiciones iniciales
    s_B = W.FloatSlider(value=30, min=1, max=300, step=1, description='B imán (mT)', style=estilo, continuous_update=False, layout=ancho)
    s_b = W.FloatSlider(value=-5.301, min=-7.3, max=-4.3, step=0.02, description='log₁₀ fricción b', style=estilo, continuous_update=False, layout=ancho)
    s_d = W.FloatSlider(value=0.5, min=0.1, max=1.5, step=0.05, description='Alambre d (mm)', style=estilo, continuous_update=False, layout=ancho)
    s_r = W.FloatSlider(value=1.5, min=1, max=4, step=0.5, description='Radio r (cm)', style=estilo, continuous_update=False, layout=ancho)
    s_R = W.FloatSlider(value=0.4, min=0.05, max=5, step=0.05, description='R externa (Ω)', style=estilo, continuous_update=False, layout=ancho)
    s_ts = W.FloatSlider(value=1.5, min=0.2, max=6, step=0.1, description='Duración (s)', style=estilo, continuous_update=False, layout=ancho)
    s_th0 = W.FloatSlider(value=0.4, min=0, max=6.28, step=0.05, description='θ inicial (rad)', style=estilo, continuous_update=False, layout=ancho)
    s_om0 = W.FloatSlider(value=0, min=0, max=100, step=1, description='ω inicial (rad/s)', style=estilo, continuous_update=False, layout=ancho)
    dd_mapa = W.Dropdown(options=[('Total (imán + bobina)', 'total'), ('Solo el imán', 'iman'), ('Solo la bobina (Biot-Savart)', 'bobina')],
                         value='total', description='Mapa de campo', style=estilo, layout=ancho)
    ck = {k: W.Checkbox(value=True, description=d, indent=False, layout=W.Layout(width='auto'))
          for k, d in (("B", "B"), ("mu", "μ"), ("tau", "τ"), ("F", "Fuerzas dF"), ("E", "Corriente / E"))}

    # Tiempo: deslizador y botón de reproducción
    s_t = W.IntSlider(value=700, min=0, max=1000, description='Instante', style=estilo, continuous_update=False,
                      layout=W.Layout(width='520px'), readout=False)
    play = W.Play(value=700, min=0, max=1000, step=10, interval=350, description='Reproducir')
    W.jslink((play, 'value'), (s_t, 'value'))
    lbl_t = W.HTML()

    # Resultados: alertas, gráficas, animación y cálculos 
    out_alert = W.HTML()
    out_fig = W.Output()
    out_anim = W.Output()
    acordeon = W.Accordion(children=[W.HTML() for _ in range(8)])
    out_tabla = W.HTML()

    def leer_params() -> Params:
        return Params(N=int(t_N.value), V=float(t_V.value), B=s_B.value * 1e-3, b=10 ** s_b.value,
                      d=s_d.value * 1e-3, r=s_r.value * 1e-2, Rext=s_R.value)

    def indice() -> int:
        n = len(est['sim']['t']) - 1
        return int(round(s_t.value / 1000 * n))

    # Se redibujan las gráficas cuando se cambian los valores inciales 
    def redibujar(*_):
        if est['sim'] is None:
            return
        mostrar = tuple(k for k, c in ck.items() if c.value)
        k = indice()
        fig = construir_figura(est['p'], est['sim'], k, dd_mapa.value, mostrar)
        with out_fig:
            clear_output(wait=True)
            display(fig)
        plt.close(fig)
        lbl_t.value = f"<span style='color:#666'>Instante mostrado: t = {est['sim']['t'][k]:.3f} s de {est['sim']['t'][-1]:.2f} s</span>"

    # Se ejectua cada vez que se cambia un valor de entrada: recalcula la simulación y los cálculos.
    def recalcular(*_):
        if est['bloqueo']:
            return
        p = leer_params()
        est['p'] = p
        est['sim'] = simular(p, t_total=s_ts.value, theta0=s_th0.value, omega0=s_om0.value)
        for (titulo, html), i in zip(bloques_calculo(p, est['sim']), range(8)):
            acordeon.children[i].value = html
            acordeon.set_title(i, titulo)
        out_alert.value = alertas_html(p, est['sim'])
        out_tabla.value = tabla_escenarios(p, float(est['sim']['T'][-1]))
        redibujar()

    # Evitar bucles infinitos: cuando se cambia un valor de entrada, se actualiza el otro sin disparar otra vez la función.
    def sync_N(cambio):
        if est['bloqueo']:
            return
        est['bloqueo'] = True
        v = max(1, min(500, int(cambio['new'] if cambio['new'] else 1)))
        t_N.value = v
        s_N.value = min(200, v)
        est['bloqueo'] = False
        recalcular()

    def sync_V(cambio):
        if est['bloqueo']:
            return
        est['bloqueo'] = True
        v = max(0.0, min(100.0, float(cambio['new'])))
        t_V.value = v
        s_V.value = min(24.0, v)
        est['bloqueo'] = False
        recalcular()

    t_N.observe(sync_N, names='value'); s_N.observe(sync_N, names='value')
    t_V.observe(sync_V, names='value'); s_V.observe(sync_V, names='value')
    for w in (s_B, s_b, s_d, s_r, s_R, s_ts, s_th0, s_om0):
        w.observe(recalcular, names='value')
    for w in [s_t, dd_mapa] + list(ck.values()):
        w.observe(redibujar, names='value')


    # Botones de ejemplo: ponen de una vez un número de espiras y un voltaje.
    def preset(n, v):
        def _f(_b):
            est['bloqueo'] = True
            t_N.value, s_N.value, t_V.value, s_V.value = n, min(200, n), v, min(24.0, v)
            est['bloqueo'] = False
            recalcular()
        return _f
    botones = []
    for etiqueta, n, v in (("Tu motor · 6 esp. · 9 V", 6, 9.0), ("Pila AA · 1,5 V · 10 esp.", 10, 1.5), ("Pila 9 V · 20 esp.", 20, 9.0),
                           ("12 V · 5 esp.", 5, 12.0), ("100 esp. · 12 V", 100, 12.0), ("1 espira · 1,5 V", 1, 1.5)):
        bt = W.Button(description=etiqueta, layout=W.Layout(width='auto'))
        bt.on_click(preset(n, v))
        botones.append(bt)

#Animación: genera un gif del giro a partir del instante elegido.
    bt_anim = W.Button(description='🎞 Generar animación (tarda ~30 s)', button_style='info', layout=W.Layout(width='auto'))

    # Crea una animación del giro a partir del instante elegido.
    def generar_animacion(_b):
        with out_anim:
            clear_output(wait=True)
            print("Generando animación…")
        p, sim = est['p'], est['sim']
        g, t = sim['g'], sim['t']
        k0 = indice()
        om_v = max(abs(sim['om'][k0]), 1.0)
        paso_t = min(0.02, 0.35 / om_v)                 # ≤ 0,35 rad por cuadro: se ve el giro sin «saltos»
        pasos = max(1, int(round(paso_t / (t[1] - t[0]))))
        idx = list(range(k0, len(t), pasos))[:60]
        if len(idx) < 20:                               # si quedan pocos cuadros, retrocede
            idx = list(range(max(0, len(t) - 1 - 59 * pasos), len(t), pasos))[:60]
        fig = plt.figure(figsize=(12, 6), dpi=70)
        gs = GridSpec(2, 2, figure=fig, width_ratios=[1.3, 1], hspace=0.4, wspace=0.18, left=0.04, right=0.98, top=0.93, bottom=0.08)
        ax3 = fig.add_subplot(gs[:, 0], projection='3d')
        ax_i = fig.add_subplot(gs[0, 1]); ax_w = fig.add_subplot(gs[1, 1])
        mostrar = tuple(k for k, c in ck.items() if c.value)
        a, b = t[idx[0]], t[idx[-1]]
        mm = (t >= a) & (t <= b)
        ax_i.plot(t[mm], sim['I'][mm], color='#e0524e', lw=1.3); ax_i.set_title("Corriente I(t)", fontsize=9)
        ax_w.plot(t[mm], sim['om'][mm], color='#9b6fd0', lw=1.3); ax_w.set_title("Velocidad ω(t)", fontsize=9)
        v1 = ax_i.axvline(a, color='tab:blue'); v2 = ax_w.axvline(a, color='tab:blue')
        for ax_ in (ax_i, ax_w):
            ax_.grid(alpha=0.3); ax_.tick_params(labelsize=7); ax_.set_xlim(a, b)

        def cuadro(j):
            kk = idx[j]
            escena3d(ax3, p, g, estado_en(sim, kk), mostrar)
            v1.set_xdata([t[kk], t[kk]]); v2.set_xdata([t[kk], t[kk]])
            return ()
        anim = FuncAnimation(fig, cuadro, frames=len(idx), interval=60, blit=False)
        html = anim.to_jshtml(fps=15)
        plt.close(fig)
        with out_anim:
            clear_output(wait=True)
            display(HTML(html))
    bt_anim.on_click(generar_animacion)

    
    entradas = W.VBox([
        W.HBox([t_N, s_N]), W.HBox([t_V, s_V]),
        W.HBox([W.Label("Ejemplos:")] + botones, layout=W.Layout(flex_flow='row wrap')),
    ])
    avanzado = W.Accordion(children=[W.VBox([W.HBox([W.VBox([s_B, s_b, s_d, s_r]), W.VBox([s_R, s_ts, s_th0, s_om0])]),
                                             W.HBox([dd_mapa] + list(ck.values()), layout=W.Layout(flex_flow='row wrap'))])])
    avanzado.set_title(0, 'Otros parámetros (imán, fricción, alambre, resistencia externa, condiciones iniciales, qué dibujar)')
    avanzado.selected_index = None
    tiempo = W.HBox([play, s_t, lbl_t])
    calc = W.VBox([W.HTML("<h3>Cálculos paso a paso con tus valores</h3>"), acordeon,
                   W.HTML("<h3>¿Qué pasa si cambio las espiras o el voltaje?</h3>"), out_tabla,
                   W.HTML("<div style='color:#555;font-size:90%;max-width:900px'>Esta tabla compara el motor con respecto a otras condiciones: el doble de espiras, la mitad, el doble de voltaje, o ambos. Los valores son los que alcanza el motor cuando ya lleva un rato girando y su velocidad se estabiliza. La fila azul es lo que se tiene ahora; las demás cambian solo las espiras o el voltaje y mantienen lo demás igual.<br><br>I₀ es la corriente al arrancar, ω_ss es la velocidad final, «P Joule» es la potencia que se pierde como calor y T_ss es la temperatura final del alambre (en rojo si pasa de 155 °C, y el esmalte se daña). Si aparece «generador», el motor gira tan rápido que, en parte de cada vuelta, el imán produce más voltaje que la fuente; el valor de esa fila es solo aproximado.</div>")])
    panel = W.VBox([W.HTML("<h2>Simulador interactivo de motor casero</h2>"), entradas, avanzado, out_alert, tiempo,
                    out_fig, W.HBox([bt_anim]), out_anim, calc])
    display(panel)
    recalcular()
    return panel


_panel = lanzar()
```

### 8.4 Análisis
- **Más espiras:** aumenta $K=NAB$, por lo que crece el torque y también la fem inducida; la velocidad terminal casi se duplica al pasar de 6 a 12 espiras. La corriente baja un poco porque el alambre más largo tiene más resistencia.
- **Más voltaje:** la corriente, el torque y la velocidad terminal crecen casi en proporción (al duplicar $V$, $\omega_{ss}$ pasa de 162 a 325 rad/s).
- **Fuerza electromotriz:** con estos valores la fem inducida es muy pequeña frente a $V$ (≈ 0,2 %). Por eso la corriente la limita casi solo la resistencia del circuito, y no el efecto de frenado de la fem. La fem aumenta con $N$, con $B$ y con la velocidad.
- **Energía y calor:** casi toda la potencia de la fuente se disipa como calor por efecto Joule; al duplicar el voltaje, el calor se multiplica por 4 (de 90 W a 360 W), porque depende de $I^2$. El rendimiento mecánico es muy bajo (0,15 %).
- **Temperatura:** si se mantuviera constante la corriente de 20 A, la temperatura de equilibrio estimada sería muy superior al límite del esmalte (155 °C). En la simulación de 1,5 s el alambre llega solo a unos 60 °C, pero un funcionamiento prolongado dañaría el esmalte.

### 8.5 Limitaciones del modelo y fuentes de error
- El campo del imán se supuso uniforme y de 30 mT; el valor real puede ser distinto y cambia directamente el torque y la fem.
- La fuente se modeló ideal con una resistencia externa de 0,4 Ω. Una batería o un Arduino reales entregan menos corriente de la calculada (20 A), por lo que la corriente y el torque reales serían menores.
- La fricción ($5\times10^{-6}$ N·m·s), la inercia del eje y el coeficiente de convección son valores estimados.
- El contacto eléctrico entre los clips y el alambre no es perfecto: produce resistencia de contacto, chispas y pérdidas que el modelo no incluye.
- La bobina real puede no estar perfectamente balanceada ni centrada sobre el imán.

## 9. Conclusiones
- Se construyó un motor de torque magnético con una bobina, un imán y un conmutador de media vuelta, y se comprobó que, tras un impulso inicial, sigue girando por sí solo.
- Se aplicaron las leyes del electromagnetismo (Biot-Savart, Lorentz, Faraday-Lenz, torque sobre un dipolo magnético y efecto Joule) para calcular el campo magnético, el campo eléctrico, la fuerza electromotriz, las fuerzas y el torque que actúan en el sistema.
- El simulador interactivo permitió ver que aumentar el número de espiras o el voltaje incrementa el torque y la velocidad final, pero también la corriente, el calor disipado y la temperatura del alambre.
- La fuerza electromotriz inducida se opone al voltaje de la fuente (ley de Lenz); con los valores usados resultó pequeña, por lo que la corriente quedó limitada sobre todo por la resistencia del circuito.
- El motor casero tiene un rendimiento mecánico muy bajo, porque casi toda la energía eléctrica se disipa como calor y el conmutador deja el circuito abierto media vuelta.
- Como mejoras futuras se proponen medir el campo real del imán y la velocidad de giro para validar el modelo, y probar otros diámetros de alambre y otras resistencias para reducir el calentamiento.
